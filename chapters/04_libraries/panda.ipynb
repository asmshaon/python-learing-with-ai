```python
students = [[50, 60, 70], [80, 90, 100 ], [78, 98, 99]] 

print(students)
```

```python
print(np.mean(students, axis=1))
```

```python
# Find best student based on average marks
# give me index of best student and his average marks
avg_marks = np.mean(students, axis=1)
print(avg_marks)
best_student_index = np.argmax(avg_marks)
print(best_student_index)
best_student_avg_marks = avg_marks[best_student_index]
print(f"Best student index: {best_student_index}, Average marks: {best_student_avg_marks}")
```
